```python
import pandas as pd
import re
import nltk
import pandas as pd
```

```python
with open("rawtext.txt", "r", encoding="utf-8") as file:
    raw_text = file.read()

print("Original Text:")
print(raw_text)
```

```python
text = raw_text.lower()

print("Lowercase Text:")
print(text)
```

```python
import re
def clean_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove HTML tags
    text = re.sub(r"<.*?>", "", text)

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", "", text)

    # Remove email addresses
    text = re.sub(r"\S+@\S+", "", text)

    # Remove @mentions
    text = re.sub(r"@\w+", "", text)

    # Remove hashtags (# only)
    text = re.sub(r"#", "", text)

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    # Remove special characters
    text = re.sub(r"[^\w\s]", "", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text

# Clean the text
cleaned_text = clean_text(raw_text)

print("Cleaned Text:\n")
print(cleaned_text)
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(raw_text)

print("Sentence Tokens:\n")
for i, sentence in enumerate(sentences, 1):
    print(f"{i}. {sentence}")
```
