```python
import pandas as pd
import re
import nltk

raw_text = """
🤖 Artificial Intelligence (AI) is transforming the world in 2026!!! 🚀 AI helps in healthcare 🏥, education 📚, banking 💳, and transportation 🚗.
Visit our website: https://www.aiworld.com or https://openai.com to learn more. 🌐
For support, email us at support@example.com or admin123@gmail.com. 📧
You can also call us at +91-9876543210 or 98765-43210. ☎️
Follow us on Twitter @AIWorldOfficial and Instagram @Future_AI 😎.
Use hashtags #ArtificialIntelligence #MachineLearning #DeepLearning #AI 🤩.
<html><body><h2>Welcome to AI!</h2></body></html>
AI systems analyze 5000+ datasets every day and achieve 98.7% accuracy. 📊💯
Researchers invested $5,000,000 in AI projects during 2026. 💰
Special characters: @#$%^&*()_+-=[]{}|;:'",.<>?/`~ !!!
Extra      spaces        should      also       be     removed.
Chatbots 🤖 can answer questions instantly 😊, recommend products 🛒, translate languages 🌍, and recognize speech 🎤.
Thank you!!! ❤️✨
"""

df = pd.DataFrame({"Conversation": [raw_text]})
print("\n========== ORIGINAL TEXT ==========\n")
print(df)

print("\n========== DATASET INFORMATION ==========")
print(df.info())
print("\n========== SHAPE ==========")
print(df.shape)
print("\n========== COLUMN NAMES ==========")
print(df.columns)
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

df["Conversation"] = df["Conversation"].str.lower()

print("\n========== LOWERCASE ==========\n")
print(df)

print("\n========== re.search() ==========")
sample = df.loc[0, "Conversation"]
if re.search(r"https?://", sample):
    print("URL Found")

print("\n========== re.findall() ==========")
numbers = re.findall(r"\d+", sample)
print("Numbers Found:", numbers)

print("\n========== re.split() ==========")
words = re.split(r"\s+", sample)
print(words)

def clean_text(text):
    text = re.sub(r"<.*?>", "", text)
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"\S+@\S+", "", text)
    text = re.sub(r"@\w+", "", text)
    text = re.sub(r"#", "", text)
    text = re.sub(r"\d+", "", text)
    text = re.sub(r"[^\w\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text

df["Clean_Conversation"] = df["Conversation"].apply(clean_text)

print("\n========== CLEANED DATA ==========\n")
print(df)
df.to_csv("cleaned_data.csv", index=False)

print("\n✅ Cleaned dataset saved successfully.")
print("\n========== BEFORE & AFTER ==========\n")

print("Original:")
print(df.loc[0, "Conversation"])

print("\nCleaned:")
print(df.loc[0, "Clean_Conversation"])
```
